```python

from google.colab import drive
drive.mount('/content/drive')

IMAGE_DIR = '/content/drive/My Drive/Graduate/Craters0727'
IMAGE_OUTPUT_DIR = '/content/drive/My Drive/Graduate/Craters0727low'
os.makedirs(IMAGE_OUTPUT_DIR, exist_ok=True)
```

```python
import cv2
import glob
import os

path = glob.glob(os.path.join(IMAGE_DIR, '*.tif'))
path = sorted(path, key=os.path.getmtime)

raw_images = []
for p in path:
    img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
    if img is not None:
        raw_images.append(img)
    else:
        print(f"画像の読み込みに失敗しました: {p}")

if not raw_images:
    raise RuntimeError(f"画像が見つかりませんでした。IMAGE_DIR='{IMAGE_DIR}' を確認してください.")
```

```python
def generate_low_resolution(img, file_path):
    new_size = (img.shape[1] // 2, img.shape[0] // 2)
    img_resized = cv2.resize(img, new_size, interpolation=cv2.INTER_CUBIC)

    # 渡された file_path からファイル名を取得して保存
    filename = os.path.basename(file_path)
    cv2.imwrite(os.path.join(IMAGE_OUTPUT_DIR, filename), img_resized)

# raw_images と path を zip でまとめて同時にループ処理する
for img, p in zip(raw_images, path):
    generate_low_resolution(img, p)
```
