```python
import argparse
import cv2
import glob
import os
from basicsr.archs.rrdbnet_arch import RRDBNet
from basicsr.utils.download_util import load_file_from_url

from realesrgan import RealESRGANer
from realesrgan.archs.srvgg_arch import SRVGGNetCompact


def main():
    """Real-ESRGANの推論用デモスクリプト"""

    # 1. コマンドライン引数の設定
    # 入出力パス、モデルの選択、タイルサイズ、顔補正の有無などを設定可能にする
    parser = argparse.ArgumentParser()
    parser.add_argument('-i', '--input', type=str, default='inputs', help='入力画像またはフォルダのパス')
    parser.add_argument(
        '-n',
        '--model_name',
        type=str,
        default='RealESRGAN_x4plus',
        help=('モデル名: RealESRGAN_x4plus | RealESRNet_x4plus | RealESRGAN_x4plus_anime_6B | '
              'RealESRGAN_x2plus | realesr-animevideov3 | realesr-general-x4v3'))
    parser.add_argument('-o', '--output', type=str, default='results', help='出力フォルダ')
    parser.add_argument(
        '-dn',
        '--denoise_strength',
        type=float,
        default=0.5,
        help=('ノイズ除去強度(0-1)。realesr-general-x4v3モデルでのみ使用'))
    parser.add_argument('-s', '--outscale', type=float, default=4, help='最終的なアップスケール倍率')
    parser.add_argument('--model_path', type=str, default=None, help='[任意] モデルファイルへのパス')
    parser.add_argument('--suffix', type=str, default='out', help='出力画像の接尾辞')
    parser.add_argument('-t', '--tile', type=int, default=0, help='タイルサイズ(0でタイル分割なし、メモリ不足時に使用)')
    parser.add_argument('--tile_pad', type=int, default=10, help='タイル境界のパディング')
    parser.add_argument('--pre_pad', type=int, default=0, help='画像周辺の事前パディング')
    parser.add_argument('--face_enhance', action='store_true', help='GFPGANを用いた顔補正機能の有効化')
    parser.add_argument('--fp32', action='store_true', help='推論時にfp32精度を使用(デフォルトはfp16)')
    parser.add_argument(
        '--alpha_upsampler',
        type=str,
        default='realesrgan',
        help='アルファチャンネルのアップサンプラー選択: realesrgan | bicubic')
    parser.add_argument(
        '--ext',
        type=str,
        default='auto',
        help='出力画像の拡張子: auto | jpg | png')
    parser.add_argument(
        '-g', '--gpu-id', type=int, default=None, help='使用するGPU ID(単一または複数)')

    args = parser.parse_args()

    # 2. モデルの初期化
    # 選択されたモデル名に応じて、適切なアーキテクチャ(RRDBNetまたはSRVGGNet)を準備
    args.model_name = args.model_name.split('.')[0]
    if args.model_name == 'RealESRGAN_x4plus':
        model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=23, num_grow_ch=32, scale=4)
        netscale = 4
        file_url = ['https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth']
    elif args.model_name == 'RealESRNet_x4plus':
        model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=23, num_grow_ch=32, scale=4)
        netscale = 4
        file_url = ['https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.1/RealESRNet_x4plus.pth']
    elif args.model_name == 'RealESRGAN_x4plus_anime_6B':
        model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=6, num_grow_ch=32, scale=4)
        netscale = 4
        file_url = ['https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.2.4/RealESRGAN_x4plus_anime_6B.pth']
    elif args.model_name == 'RealESRGAN_x2plus':
        model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=23, num_grow_ch=32, scale=2)
        netscale = 2
        file_url = ['https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.1/RealESRGAN_x2plus.pth']
    elif args.model_name == 'realesr-animevideov3':
        model = SRVGGNetCompact(num_in_ch=3, num_out_ch=3, num_feat=64, num_conv=16, upscale=4, act_type='prelu')
        netscale = 4
        file_url = ['https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.5.0/realesr-animevideov3.pth']
    elif args.model_name == 'realesr-general-x4v3':
        model = SRVGGNetCompact(num_in_ch=3, num_out_ch=3, num_feat=64, num_conv=32, upscale=4, act_type='prelu')
        netscale = 4
        file_url = [
            'https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.5.0/realesr-general-wdn-x4v3.pth',
            'https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.5.0/realesr-general-x4v3.pth'
        ]

    # 3. モデル重みファイルの読み込み(ダウンロード)
    # ローカルにない場合はGitHubからダウンロード
    if args.model_path is not None:
        model_path = args.model_path
    else:
        model_path = os.path.join('weights', args.model_name + '.pth')
        if not os.path.isfile(model_path):

            ROOT_DIR = os.getcwd()
            for url in file_url:
                model_path = load_file_from_url(
                    url=url, model_dir=os.path.join(ROOT_DIR, 'weights'), progress=True, file_name=None)

    # 4. ノイズ除去強度(DNI)の適用
    # 汎用モデルのみ、ノイズ除去の強さを調整可能
    dni_weight = None
    if args.model_name == 'realesr-general-x4v3' and args.denoise_strength != 1:
        wdn_model_path = model_path.replace('realesr-general-x4v3', 'realesr-general-wdn-x4v3')
        model_path = [model_path, wdn_model_path]
        dni_weight = [args.denoise_strength, 1 - args.denoise_strength]

    # 5. 推論器(RealESRGANer)のインスタンス化
    upsampler = RealESRGANer(
        scale=netscale,
        model_path=model_path,
        dni_weight=dni_weight,
        model=model,
        tile=args.tile,
        tile_pad=args.tile_pad,
        pre_pad=args.pre_pad,
        half=not args.fp32,
        gpu_id=args.gpu_id)

    # 6. 顔補正(GFPGAN)の準備
    if args.face_enhance:
        from gfpgan import GFPGANer
        face_enhancer = GFPGANer(
            model_path='https://github.com/TencentARC/GFPGAN/releases/download/v1.3.0/GFPGANv1.3.pth',
            upscale=args.outscale,
            arch='clean',
            channel_multiplier=2,
            bg_upsampler=upsampler)

    os.makedirs(args.output, exist_ok=True)

    # 7. 入力ファイルのリスト化
    if os.path.isfile(args.input):
        paths = [args.input]
    else:
        paths = sorted(glob.glob(os.path.join(args.input, '*')))

    # 8. 推論ループの実行
    for idx, path in enumerate(paths):
        imgname, extension = os.path.splitext(os.path.basename(path))
        print('Testing', idx, imgname)

        img = cv2.imread(path, cv2.IMREAD_UNCHANGED)
        if len(img.shape) == 3 and img.shape[2] == 4:
            img_mode = 'RGBA'
        else:
            img_mode = None

        try:
            # 顔補正ありの場合と、通常アップスケールの場合で分岐
            if args.face_enhance:
                _, _, output = face_enhancer.enhance(img, has_aligned=False, only_center_face=False, paste_back=True)
            else:
                output, _ = upsampler.enhance(img, outscale=args.outscale)
        except RuntimeError as error:
            print('Error', error)
            print('CUDA out of memory発生時は --tile の値を小さくしてください。')
        else:
            # 9. 結果の保存
            if args.ext == 'auto':
                extension = extension[1:]
            else:
                extension = args.ext
            if img_mode == 'RGBA':
                extension = 'png' # RGBAはPNG形式で保存
            if args.suffix == '':
                save_path = os.path.join(args.output, f'{imgname}.{extension}')
            else:
                save_path = os.path.join(args.output, f'{imgname}_{args.suffix}.{extension}')
            cv2.imwrite(save_path, output)

if __name__ == '__main__':
    # 変更前: main()
    # 変更後: main() を呼び出さず、引数を直接指定するか、リストを空にする
    import sys
    sys.argv = ['inference_realesrgan.py', '-n', 'RealESRGAN_x4plus', '-i', 'inputs', '--outscale', '4', '--face_enhance']
    main()
```

```python
import cv2
import os
import glob
from basicsr.archs.rrdbnet_arch import RRDBNet
from realesrgan import RealESRGANer

def run_inference():
    # --- 設定エリア ---
    INPUT_FOLDER = 'input2'
    OUTPUT_FOLDER = 'result'
    # アニメ用モデル名に変更
    MODEL_NAME = 'RealESRGAN_x4plus_anime_6B'
    OUTSCALE = 4
    FACE_ENHANCE = False # アニメ画像には通常GFPGANは不要なのでFalse推奨
    # ------------------

    # 【重要】アニメモデル(6B)の構造定義
    # num_block=6 にするのがアニメモデルの特徴です
    model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=6, num_grow_ch=32, scale=4)

    # モデルのダウンロード先パス
    os.makedirs('experiments/pretrained_models', exist_ok=True)
    model_path = os.path.join('experiments/pretrained_models', MODEL_NAME + '.pth')

    # まだモデルがない場合はダウンロードする処理を追加
    if not os.path.exists(model_path):
        print(f"{MODEL_NAME}をダウンロード中...")
        import torch
        url = 'https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.2.4/RealESRGAN_x4plus_anime_6B.pth'
        torch.hub.download_url_to_file(url, model_path)

    # 推論器の初期化
    upsampler = RealESRGANer(
        scale=4,
        model_path=model_path,
        model=model,
        tile=0,
        half=True)

    # 顔補正（必要なら）
    if FACE_ENHANCE:
        from gfpgan import GFPGANer
        face_enhancer = GFPGANer(
            model_path='https://github.com/TencentARC/GFPGAN/releases/download/v1.3.0/GFPGANv1.3.pth',
            upscale=OUTSCALE,
            arch='clean',
            channel_multiplier=2,
            bg_upsampler=upsampler)

    os.makedirs(OUTPUT_FOLDER, exist_ok=True)

    # 処理ループ
    paths = sorted(glob.glob(os.path.join(INPUT_FOLDER, '*')))
    for path in paths:
        imgname = os.path.basename(path)
        # 画像ファイル以外（隠しファイル等）を除外するチェック
        if not imgname.lower().endswith(('.png', '.jpg', '.jpeg', '.webp')):
            continue

        print(f'処理中: {imgname}')
        img = cv2.imread(path, cv2.IMREAD_UNCHANGED)
        # img = cv2.imread(...) の直後に追加
        img = cv2.fastNlMeansDenoisingColored(img, None, 10, 10, 7, 21)

        try:
            if FACE_ENHANCE:
                _, _, output = face_enhancer.enhance(img, has_aligned=False, only_center_face=False, paste_back=True)
            else:
                output, _ = upsampler.enhance(img, outscale=OUTSCALE)

            cv2.imwrite(os.path.join(OUTPUT_FOLDER, imgname), output)
        except Exception as e:
            print(f'エラーが発生しました: {e}')

    print('すべての処理が完了しました！')

run_inference()
```
